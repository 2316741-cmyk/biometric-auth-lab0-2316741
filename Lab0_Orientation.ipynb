```python
scores = [85, 92, 78, 90, 88]
print("scores:", scores)
```

```python
print("Average:", sum(scores) / len(scores))
print("Minimum:", min(scores))
print("Maximum:", max(scores))
```

```python
import matplotlib.pyplot as plt
students = ["Quiz1", "Quiz2", "Quiz3", "Quiz4", "Quiz5"]
plt.figure(figsize=(6, 4))
plt.bar(students, scores, color="steelblue")
plt.xlabel("Quiz")
plt.ylabel("score")
plt.title("My Quiz Scores")
plt.savefig("my_first_chart.png", dpi=150, bbox_inches="tight")
plt.show()
```
